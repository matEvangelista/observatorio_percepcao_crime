```python
from transformers import pipeline

ner = pipeline(
    "ner",
    model="marcosgg/bert-LARGE-pt-ner-enamex",
    tokenizer="marcosgg/bert-large-pt-ner-enamex",
    aggregation_strategy="simple"
)

result = ner("Flagrante em Vigário Geral na Zona Norte do Rio 🐠🇮🇱.")
print(result)
```

```python
result = ner("""Muitos b4ndidos em moto no Rio!
Roubo de veículo Nissan Kicks em Oswaldo Cruz. 

Ainda esculacharam o motorista.""")
print(result)
```
