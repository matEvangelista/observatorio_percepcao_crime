```python
df_isp = pd.read_csv("https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv", sep=';', encoding='latin-1')
df_isp = df_isp[(df_isp.ano == 2025) & (df_isp.munic == 'Rio de Janeiro')]
#df_isp = df_isp.merge(cisps_cidade, how='inner', on='cisp')
df_isp
```

```python
colunas_uteis = [
    # Identificadores base
    "cisp",
    "ano",
    "mes",
    
    # Letalidade violenta principal
    "hom_doloso",
    "lesao_corp_morte",
    "latrocinio",
    "hom_por_interv_policial",
    "tentat_hom",
    
    # Contra pessoa
    "lesao_corp_dolosa",
    "estupro",
    "hom_culposo",
    "lesao_corp_culposa",
    "ameaca",
    
    # Patrimônio agregados + principais desagregados
    "roubo_transeunte",
    "roubo_celular",      # Mais frequente
    "roubo_em_coletivo",
    "roubo_veiculo",
    "roubo_carga",
    "roubo_rua",
    "roubo_residencia",
    "roubo_banco",
    "roubo_comercio",
    "roubo_cx_eletronico",
    "roubo_conducao_saque",
    "roubo_apos_saque",
    "roubo_bicicleta",
    "outros_roubos",

    "furto_veiculos",
    "furto_celular",
    'furto_transeunte',
    'furto_coletivo',
    'furto_bicicleta',
    "outros_furtos",
    
    # Outros contra patrimônio/pessoa
    "sequestro",
    "sequestro_relampago",
    "extorsao",
    "estelionato",
    
    # Drogas
    "posse_drogas",
    "trafico_drogas",
    
    # Indicadores adicionais essenciais
    "pessoas_desaparecidas",
    "encontro_cadaver",
    "pol_militares_mortos_serv",
    "pol_civis_mortos_serv",
    "registro_ocorrencias"
]
df_uteis = df_isp[colunas_uteis]
df_uteis = df_isp[colunas_uteis]
df_uteis
```

```python
df_uteis.loc[:, "policiais_mortos"] = (
    df_isp["pol_militares_mortos_serv"] +
    df_isp["pol_civis_mortos_serv"]
)
df_uteis = df_uteis.drop(columns=['pol_militares_mortos_serv', 'pol_civis_mortos_serv'])
df_uteis
```

```python
df_uteis = df_uteis.rename(columns={'hom_doloso': 'homicidio',
                                    'hom_por_interv_policial': 'hom_por_intervencao_policial'})
#df_uteis = df_uteis.drop(columns=['cisp'])
df_uteis
```

```python
colunas_para_somar = list(df_uteis.columns)[3:]

# 2. Filtrar apenas as linhas das CISPs 12 e 13
df_subset = df_uteis[df_uteis['cisp'].isin([12, 13])].copy()

# 3. Agrupar por Ano e Mês e somar os crimes
# Isso funde os dados da 12 e da 13 mantendo a linha do tempo correta
df_soma = df_subset.groupby(['ano', 'mes'])[colunas_para_somar].sum().reset_index()

# 4. Atribuir o id 13 para essas linhas somadas
df_soma['cisp'] = 13

# 5. Remover as linhas originais da 12 e da 13 do dataframe principal
df_uteis = df_uteis[~df_uteis['cisp'].isin([12, 13])]

# 6. Adicionar as novas linhas somadas de volta ao dataframe
df_uteis = pd.concat([df_uteis, df_soma], ignore_index=True)

# Opcional: Reordenar para manter organizado
df_uteis = df_uteis.sort_values(by=['ano', 'mes', 'cisp'])
```
